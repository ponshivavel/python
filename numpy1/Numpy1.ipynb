```python
# 3. Array initialization methods
```

```python
# Zero Array
zero_arr = np.zeros((2,3))
print(zero_arr)
```

```python
# One Array
One_arr = np.ones((3,3))
print(One_arr)
```

```python
Full_arr = np.full((2,3),7)
print(Full_arr)
```

```python
Full_arr = np.full((3,3),4)
print(Full_arr)
```

```python
# Identity Methods/Matrix/3x3/Square Matrix

id_arr = np.eye(3)
print(id_arr)
```

```python
# Empty Array
print(np.empty(2))
```

```python
# Random values array -float
r_arr = np.random.rand(3,2)
print(r_arr)
```

```python
# 4. array Indexing and slicing
```

```python
a =np.array([1,2,3,4,5])
print(a)
```

```python
print(a[0])
```

```python
print(a[3])
```

```python
a=np.array([3,5,6,7,3,4,5,6,9,1,2])
print(a)
# Slicing:
# array[start:stop:step]
# stop - index, that is not included in output.
```

```python
print(a[0:3]) # First 3 elements
```

```python
print(a[-6:]) # Last 3 elements
```

```python
print(a[1:8:2])
```
