```python
# 10. Concatenate and Merge Dataframe (JOINs)
```

```python
df1 = pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
df1
```

```python
df2 = pd.DataFrame({'ID':[1,2,3,4],'Score':[98,45,33,25]})
df2
```

```python
pd.concat([df1,df2],axis=0) # Shows vertically
```

```python
pd.concat([df1,df2],axis=1) # Shows rows wise
```

```python
# Merge = join
pd.merge(df1,df2,how ='inner',on ='ID')
```
