```python
import matplotlib.pyplot as plt

# pyplot is a collection of functions that make matplotlib as MATLAB
```

```python
# Data

x = [1,2,3,4,5]
y = [10,25,30,35,52]

plt.plot(x,y) # line chart
plt.show() # Display chart in jupyter notebook
```

```python
# 2 Customize Chart
```

```python
plt.figure(figsize=(5,3))

plt.plot(x,y,color ='yellow',marker='o',linestyle='dashed',linewidth=2,markersize=12)

plt.title(" Rupees ")
plt.xlabel(" Price movement ")
plt.ylabel(" volume movement ")

plt.show()
```

```python
# 3. Advanced line chart - multiple lines & legends
```

```python
x = [1,2,3,4,5]
y1 = [100,120,220,340,450]
y2 = [200,305,310,410,500]

plt.plot(x,y1,label=' Liq 2024') # Ploting y1 data
plt.plot(x,y2,label=' vol 2025') # Ploting y2 data

plt.title(" BTC DATA ")
plt.xlabel(" Liqudity movement ")
plt.ylabel(" volume movement ")

plt.legend()
plt.show()
```

```python
# Bar chart
```

```python
# Data

x = [1,2,3,4,5]
y = [10,25,34,4,52]
plt.bar(x,y)
plt.title('Bar Chart')
plt.show()
```

```python
# Histogram
```

```python
# Data
import random
data = [random.randint(1,80)for _ in range(900)]

# Used for distribution analysis.
plt.hist(data,bins=80)
plt.title("Histogram")
plt.show()
```

```python
# Pie Chart - Used to show part to whole relationship/data.
```

```python
# Data
Categories=['A','B','C','D','E']
sales=[10, 20, 32, 45, 58]
plt.pie(sales, labels=Categories, autopct = '%1.1f%%',startangle=90)
plt.title("Pie Chart")
plt.show()
```

```python
# Scatter plot - Used to find relationship b/w variables.
```

```python
y1 = [100,120,220,340,400]
y2 = [200,305,310,410,500]
plt.scatter(y1,y2)
plt.title("Scatter Chart")
plt.show()
```

```python
# Subplots - used to show multiple chart in one figure .
```

```python
# DATA-1 - Bar chart

Categories=['Mon','Tue','Web','Thur','Fri']
sales=[10, 20, 32, 45, 58]

# DATA-2 - Scatter plot

y1 = [10,12,22,34,40]
y2 = [20,30,31,41,50]

plt.figure(figsize=(10,5))

# First plot - Bar chart
plt.subplot(1,2,1) # Row , Column , Position.
plt.bar(Categories,sales)
plt.title("Weekly Sales")
plt.xlabel("Week Days")
plt.ylabel("Sales")


# Second plot - scatter chart

plt.subplot(1,2,2) # Rows,column,Position.
plt.scatter(y1,y2)
plt.title("User Example")
plt.xlabel("User1")
plt.ylabel("User2")


plt.show()
```

```python
# Create df
import pandas as pd

# Data

data = {
    'Month':['Jan','Feb','Mar','Apr'],
    'Sales':[12000,11000,13000,25000]
}
df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title("Pandas")
plt.show()
```

```python
# 10 Save Plot/Charts
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title("Pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("Sales.png")
plt.show()

```
