```python
# 5. Array Reshaping and Flattening
```

```python
arr1 = np.array([1,2,3,4])
print(arr1)
```

```python
reshaped = arr1.reshape((4,1))
print(reshaped)
```

```python
reshaped2 = arr1.reshape((4,1))
print(reshaped2)
```

```python
print(reshaped2.flatten())
```

```python
# 6 Array_Stacking_and_Spliting
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.vstack((a,b))) # vertical stacking - row wise
```

```python
print(np.hstack((a,b))) # Horizontal stacking - colomn wise
```

```python
c = np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
split = np.hsplit(c,3)

for s in split:
    print(s)
```

```python
split = np.vsplit(c,2)

for s in split:
    print(s)
```

```python
# Mathematical Operation on Array
```

```python
a = np.array([1,2,3])
print(a)
```

```python
print(a+10)
print(a-3)
print(a*2)
print(a/1)
```

```python
b = np.array([1,4,9])
print(b)
```

```python
print(np.square(b))
print(np.sqrt(b))
```

```python
a = np.array([[[1,2,3],
               [4,5,6],
               [8,9,7]]])
b = np.array([[[2,3,4],
               [8,7,9],
               [4,5,6]]])
print(a)
print(b)
```

```python
print(a+b)
print(a-b)
print(a*b)
```

```python
print(a.T)
```

```python
# Statistical Functions
```

```python
a = np.array([[1,2,3],[4,5,6]])
print(a)
```

```python
print(np.sum(a))
print(np.mean(a))
print(np.min(a))
print(np.max(a))
```

```python
a = np.array([[[1,2,3],
               [4,5,6],
               [8,9,7]]])
b = np.array([[[2,3,4],
               [8,7,9],
               [4,5,6]]])
print(a)
print(b)
```

```python
print(a == b)
```

```python
print(np.array_equal(a,b))
```
